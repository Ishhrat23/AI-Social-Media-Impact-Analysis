## Conclusions
- Daily social media use shows the strongest negative relationship with mental health (-0.40), and is also linked to lower physical health (-0.31) and less sleep (-0.30).
- AI tool usage has only weak relationships with sleep and mental health (-0.14 and -0.11), and almost none with physical health or activity.
- Physical activity is most strongly linked to physical health (0.42), and sleep is linked to better mental health (0.31).
- Age shows no meaningful correlation with any variable.
- All correlations are weak to moderate, so these are tendencies, not strong predictors.

*Note: Correlation does not imply causation. These results show associations only, not that one factor causes another.*
